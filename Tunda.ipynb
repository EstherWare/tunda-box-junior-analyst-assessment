```python
# Cell 2: Question S1 - Orders by Status
# We loop through all order records and count how many orders exist for each status.

status_counts = {}

for o in orders:
    st = o["status"]
    status_counts[st] = status_counts.get(st, 0) + 1

print("--- Question S1 Results ---")
for status_name, count in status_counts.items():
    print(f"Status '{status_name}': {count} orders")
```

```python
# Cell 3: Question S2 - Revenue by City
# Clean city name strings and map customer IDs to cleaned cities
def clean_city(name):
    return name.replace('"', '').strip().title()

cust_city_map = {c["customer_id"]: clean_city(c["city"]) for c in customers}

city_revenue = {}

for o in orders:
    if o["status"] == "completed" and o["amount_kes"] is not None:
        cid = o["customer_id"]
        city = cust_city_map.get(cid, "Unknown")
        city_revenue[city] = city_revenue.get(city, 0) + o["amount_kes"]

print("--- Question S2 Results ---")
for city, rev in sorted(city_revenue.items(), key=lambda x: x[1], reverse=True):
    print(f"{city}: KES {rev:,}")
```

```python
# Cell 4: Question S3 - Customers Who Never Ordered
# Extract set of customer IDs who placed orders and filter customers missing from that set

ordered_customer_ids = {o["customer_id"] for o in orders}

never_ordered = [
    c for c in customers if c["customer_id"] not in ordered_customer_ids
]

print("--- Question S3 Results ---")
for c in never_ordered:
    print(f"Customer ID {c['customer_id']}: {c['city']} (Channel: {c['channel']})")
```

```python
# Cell 5: Question S4 - Repeat Purchase Rate
# Calculate order counts per registered customer and compute the repeat purchase rate.

customer_order_counts = {c["customer_id"]: 0 for c in customers}

for o in orders:
    cid = o["customer_id"]
    if cid in customer_order_counts:
        customer_order_counts[cid] += 1

repeat_customers = sum(1 for count in customer_order_counts.values() if count >= 2)
total_registered = len(customers)
repeat_rate = (repeat_customers / total_registered) * 100

print("--- Question S4 Results ---")
print(f"Repeat Buyers (>=2 orders): {repeat_customers}")
print(f"Total Registered Customers: {total_registered}")
print(f"Repeat Purchase Rate: {repeat_rate:.1f}%")
```

```python
# Cell 6: Question S5 - Month-on-Month Revenue & Growth
# Aggregate completed revenue by month (YYYY-MM) and compute percentage growth vs previous month.

monthly_rev = {}

for o in orders:
    if o["status"] == "completed" and o["amount_kes"] is not None:
        month = o["order_date"][:7] # Extract YYYY-MM
        monthly_rev[month] = monthly_rev.get(month, 0) + o["amount_kes"]

sorted_months = sorted(monthly_rev.keys())
prev_rev = None

print("--- Question S5 Results ---")
print(f"{'Month':<10} | {'Revenue (KES)':<15} | {'MoM Growth (%)':<15}")
print("-" * 48)

for m in sorted_months:
    curr_rev = monthly_rev[m]
    if prev_rev is None:
        growth_str = "N/A (Baseline)"
    else:
        growth = ((curr_rev - prev_rev) / prev_rev) * 100
        growth_str = f"{growth:+.1f}%"

    print(f"{m:<10} | KES {curr_rev:<11,} | {growth_str:<15}")
    prev_rev = curr_rev
```

```python
# Cell 7: Part 5 - Business Case & Simpson's Paradox Verification
# Programmatically calculate conversion rates by channel and overall for Sep vs Oct.

traffic_data = [
    {"channel": "Organic", "sep_sessions": 40000, "sep_orders": 2400, "oct_sessions": 38000, "oct_orders": 2318},
    {"channel": "Paid Social", "sep_sessions": 10000, "sep_orders": 200, "oct_sessions": 30000, "oct_orders": 630}
]

print("--- Part 5 Diagnostic Verification ---")
total_sep_sess = sum(d["sep_sessions"] for d in traffic_data)
total_sep_ord = sum(d["sep_orders"] for d in traffic_data)
total_oct_sess = sum(d["oct_sessions"] for d in traffic_data)
total_oct_ord = sum(d["oct_orders"] for d in traffic_data)

for d in traffic_data:
    sep_cr = (d["sep_orders"] / d["sep_sessions"]) * 100
    oct_cr = (d["oct_orders"] / d["oct_sessions"]) * 100
    print(f"{d['channel']}: Sep CR = {sep_cr:.1f}%, Oct CR = {oct_cr:.1f}%")

print("-" * 45)
print(f"Overall Site: Sep CR = {(total_sep_ord/total_sep_sess)*100:.1f}%, Oct CR = {(total_oct_ord/total_oct_sess)*100:.1f}%")
```
